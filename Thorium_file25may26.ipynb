```python
import numpy as np
import matplotlib.pyplot as plt

# Data points matching Figure 1
burnup = np.linspace(10, 100, 100)
cr = 1.14 - (1.14 - 1.05) * (burnup - 10) / (100 - 10)

# Plot configuration
plt.figure(figsize=(6, 5))
plt.plot(burnup, cr, color='#1f77b4', linewidth=1.5)

# Labeling and bounds matches to original image
plt.title('CR vs Burnup', fontsize=11)
plt.xlabel('Burnup (GWd/t)', fontsize=10)
plt.ylabel('Conversion Ratio (CR)', fontsize=10)
plt.xlim(5, 105)
plt.ylim(1.045, 1.145)
plt.grid(True, linestyle='-', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# python code for fiure 2
import numpy as np
import matplotlib.pyplot as plt

# Data generation for Neutron Flux vs Efficiency (Logarithmic relationship)
# Flux ranging from 1e13 to 1e15 n/cm^2/s
flux = np.linspace(1e13, 1e15, 500)
efficiency = 0.25 + 0.20 * (np.log(flux) - np.log(1e13)) / (np.log(1e15) - np.log(1e13))

# Plot configuration
plt.figure(figsize=(6, 5))
plt.plot(flux, efficiency, color='#1f77b4', linewidth=1.5)

# Labeling and bounds matching the original image
plt.title('Neutron Flux vs Efficiency', fontsize=11)
plt.xlabel('Neutron Flux (n/cm^2/s)', fontsize=10)
plt.ylabel('Efficiency', fontsize=10)

# Adjust limits and grid to match the attached figure exactly
plt.xlim(-0.05e15, 1.05e15)
plt.ylim(0.24, 0.46)
plt.yticks(np.arange(0.250, 0.451, 0.025))
plt.grid(True, linestyle='-', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate time data (e.g., 0 to 600 days of reactor operation)
time = np.linspace(0, 600, 1000)

# 2. Mathematical model for a self-sustaining thorium cycle buildup
# Starts at ~0.88 (initial startup phase before U-233 equilibrium)
# Rises and stabilizes asymptotically at 1.03 (net breeding)
breeding_ratio = 1.03 - 0.15 * np.exp(-time / 90)

# 3. Initialize the plot
plt.figure(figsize=(7.5, 5.5))
plt.plot(time, breeding_ratio, color='#2ca02c', linewidth=2.5, label='Thorium Fuel Cycle (BR)')

# 4. Add the "Self-Sustaining" threshold line at BR = 1.0
plt.axhline(y=1.0, color='#d62728', linestyle='--', linewidth=1.5, label='Break-even Threshold (BR = 1.0)')

# 5. Labels, Title, and Annotations based on your text
plt.title('Breeding Ratio vs Time', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Time (Days of Operation)', fontsize=12)
plt.ylabel('Breeding Ratio (BR)', fontsize=12)

# Add text notes directly on the graph to explain the phases
plt.text(50, 0.92, 'U-233 Buildup Phase\n(BR is increasing)', fontsize=10, color='black', bbox=dict(facecolor='white', alpha=0.6))
plt.text(320, 1.04, 'Self-Sustaining Equilibrium\n(BR stabilizes > 1.0)', fontsize=10, color='black', bbox=dict(facecolor='white', alpha=0.6))

# 6. Grid and Axis formatting
plt.grid(True, which='both', linestyle=':', linewidth=0.5, color='gray')
plt.xlim(0, 600)
plt.ylim(0.85, 1.10)
plt.legend(loc='lower right', fontsize=11)

# 7. Save and display the clean figure
plt.savefig('Thorium_Self_Sustaining_Cycle.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Apply universal crisp formatting configurations for scientific plots
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.linewidth'] = 1.0
plt.rcParams['grid.color'] = '#cbcbcb'
plt.rcParams['grid.linestyle'] = '-'
plt.rcParams['grid.linewidth'] = 0.75

# =====================================================================
# FIGURE 1: Neutron Flux vs Efficiency (image_063d4d.png)
# =====================================================================
fig1, ax1 = plt.subplots(figsize=(6.8, 5.2))

# Simulating high initial slope logarithmic/exponential asymptotic curve
# Fits data perfectly from (0.01e15, 0.25) leading up to (1.0e15, 0.45)
flux = np.linspace(0.01e15, 1.0e15, 1000)
efficiency = 0.25 + 0.20 * (1 - np.exp(-3.1 * (flux - 0.01e15) / 1e15))**0.44

ax1.plot(flux, efficiency, color='#1f77b4', linewidth=1.7)

ax1.set_title('Neutron Flux vs Efficiency', fontsize=13, pad=10)
ax1.set_xlabel('Neutron Flux (n/cm^2/s)', fontsize=11)
ax1.set_ylabel('Efficiency', fontsize=11)

# Axis limits and scientific notation configurations matching the image
ax1.set_xlim(-0.04e15, 1.05e15)
ax1.set_ylim(0.24, 0.46)
ax1.ticklabel_format(style='sci', axis='x', scilimits=(15, 15))
ax1.grid(True)

plt.tight_layout()
plt.savefig('1_neutron_flux_vs_efficiency.png', dpi=300)
plt.show()


# =====================================================================
# FIGURE 2: CR vs Burnup (image_06410a.png)
# =====================================================================
fig2, ax2 = plt.subplots(figsize=(6.8, 5.2))

# Extracted exact endpoints from linear negative slope trend line
burnup = np.array([10, 100])
cr = np.array([1.14, 1.05])

ax2.plot(burnup, cr, color='#1f77b4', linewidth=1.7)

ax2.set_title('CR vs Burnup', fontsize=13, pad=10)
ax2.set_xlabel('Burnup (GWd/t)', fontsize=11)
ax2.set_ylabel('Conversion Ratio (CR)', fontsize=11)

# Axis limits and ticks mirroring the provided attachment configuration
ax2.set_xlim(5, 105)
ax2.set_ylim(1.045, 1.145)
ax2.set_xticks([20, 40, 60, 80, 100])
ax2.grid(True)

plt.tight_layout()
plt.savefig('2_cr_vs_burnup.png', dpi=300)
plt.show()


# =====================================================================
# FIGURE 3: Breeding Ratio vs Time (image_168724.png)
# =====================================================================
fig3, ax3 = plt.subplots(figsize=(6.8, 5.2))

# Asymptotic mathematical model mirroring the 40-year thorium cycle buildup
# Starting exactly at 1.02 at t=0, approaching close to 1.12 at t=40
time_years = np.linspace(0, 40, 1000)
breeding_ratio = 1.12 - 0.10 * np.exp(-time_years / 11.2)

ax3.plot(time_years, breeding_ratio, color='#1f77b4', linewidth=1.7)

ax3.set_title('Breeding Ratio vs Time', fontsize=13, pad=10)
ax3.set_xlabel('Time (years)', fontsize=11)
ax3.set_ylabel('Breeding Ratio', fontsize=11)

# Axis limits formatting matching the final attachment
ax3.set_xlim(-2, 42)
ax3.set_ylim(1.015, 1.123)
ax3.grid(True)

plt.tight_layout()
plt.savefig('3_breeding_ratio_vs_time.png', dpi=300)
plt.show()
```
